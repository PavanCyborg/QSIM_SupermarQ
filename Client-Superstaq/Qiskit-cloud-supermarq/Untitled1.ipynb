```python
import re

def is_approximately_equal(expected_hex, got_hex, tolerance=0.01):
    expected = int(expected_hex, 16)
    got = int(got_hex, 16)
    return abs(2 * expected - got) <= tolerance * got

def check_mismatches(file_path):
    pattern = r"Expected ([0-9a-fA-F]+), Got ([0-9a-fA-F]+)"
    
    with open(file_path, 'r') as file:
        for line_num, line in enumerate(file, 1):
            match = re.search(pattern, line)
            if match:
                expected_hex, got_hex = match.groups()
                if is_approximately_equal(expected_hex, got_hex):
                    print(f"[Line {line_num}] ✅ Approx match: 2*{expected_hex} ≈ {got_hex}")
                else:
                    print(f"[Line {line_num}] ❌ No match: 2*{expected_hex} ≠ {got_hex}")

# Example usage
check_mismatches("new5.txt")
```

```python
import re

def hex_str_to_bytes(hex_str):
    if len(hex_str) % 2 != 0:
        hex_str = "0" + hex_str
    return [int(hex_str[i:i+2], 16) for i in range(0, len(hex_str), 2)]

def simulate_bytewise_double(hex_bytes):
    result = []
    carry = 0
    for byte in reversed(hex_bytes):
        doubled = (byte << 1) + carry
        result.insert(0, doubled & 0xFF)
        carry = (doubled >> 8) & 0x1
    if carry:
        result.insert(0, carry)
    return result

def is_within_one_bit_per_byte(simulated, got):
    if len(simulated) > len(got):
        got = [0] * (len(simulated) - len(got)) + got
    elif len(got) > len(simulated):
        simulated = [0] * (len(got) - len(simulated)) + simulated

    for s, g in zip(simulated, got):
        if abs(s - g) > 1:
            return False
    return True

def bytes_to_hex_str(byte_list):
    return ''.join(f"{b:02x}" for b in byte_list).lstrip('0') or '00'

def check_mismatches(file_path):
    pattern = r"Expected ([0-9a-fA-F]+), Got ([0-9a-fA-F]+)"
    
    with open(file_path, 'r') as file:
        for line_num, line in enumerate(file, 1):
            match = re.search(pattern, line)
            if match:
                expected_hex, got_hex = match.groups()
                expected_bytes = hex_str_to_bytes(expected_hex)
                got_bytes = hex_str_to_bytes(got_hex)

                simulated = simulate_bytewise_double(expected_bytes)
                simulated_hex = bytes_to_hex_str(simulated)

                if is_within_one_bit_per_byte(simulated, got_bytes):
                    print(f"[Line {line_num}] ✅ Approx match: 2*{expected_hex} = {simulated_hex} ≈ {got_hex}")
                else:
                    print(f"[Line {line_num}] ❌ No match: 2*{expected_hex} = {simulated_hex} ≠ {got_hex}")

# Example usage
check_mismatches("new 5.txt")
```
