```python
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
from qiskit import Aer

# Create a quantum circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# Apply Hadamard gate to the first qubit to create superposition
qc.h(0)

# Apply CNOT gate with the first qubit as control and second as target to entangle
qc.cx(0, 1)

# Measure both qubits
qc.measure([0, 1], [0, 1])

# Execute the circuit on a simulator
backend = Aer.get_backend('statevector_simulator')
job = backend.run(qc,shots=1024)
result = job.result()
print(result)

# Get and plot the counts (measurement outcomes)
counts = result.get_counts()
print("Measurement results:", counts)
plot_histogram(counts)
plt.show()
```

```python
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
from qiskit_aer.backends.fpga_simulator import FPGASimulator

# Create a quantum circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# Apply Hadamard gate to the first qubit to create superposition
qc.h(0)

# Apply CNOT gate with the first qubit as control and second as target to entangle
qc.cx(0, 1)

# Measure both qubits
qc.measure([0, 1], [0, 1])

# Execute the circuit on a simulator
backend = FPGASimulator() 
job = backend.run(qc)
result = job
print(result)

# Get and plot the counts (measurement outcomes)
counts = result.get_counts()
print("Measurement results:", counts)
plot_histogram(counts)
plt.show()
```
