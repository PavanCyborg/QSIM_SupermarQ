```python
from qiskit import QuantumCircuit
from qiskit_ibm_runtime import QiskitRuntimeService, SamplerV2 as Sampler
 
# Create empty circuit
example_circuit = QuantumCircuit(2)
example_circuit.measure_all()
 
# You'll need to specify the credentials when initializing QiskitRuntimeService, if they were not previously saved.
service = QiskitRuntimeService()
backend = service.least_busy(operational=True, simulator=False)

from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
 
pm = generate_preset_pass_manager(backend=backend, optimization_level=1)
isa_circuit = pm.run(example_circuit)
 
sampler = Sampler(backend)
job = sampler.run([isa_circuit])
print(f"job id: {job.job_id()}")
result = job.result()
print(result)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# === Load sheet with multiple tables ===
file_path = "SupermarQ.xlsx"
df_raw = pd.read_excel(file_path, sheet_name="Sheet1")

# === Split the DataFrame into chunks by blank rows ===
tables = []
current_table = []

for _, row in df_raw.iterrows():
    if row.isnull().all():
        if current_table:
            tables.append(pd.DataFrame(current_table))
            current_table = []
    else:
        current_table.append(row)

if current_table:  # Catch the last table
    tables.append(pd.DataFrame(current_table))

# === Process each table ===
for i, table in enumerate(tables):
    table.columns = table.iloc[0]
    table = table[1:].reset_index(drop=True)
    table = table.dropna(axis=1, how='all')  # Drop any fully empty columns

    # Convert Qubits and time columns to numeric
    table.iloc[:, 0] = pd.to_numeric(table.iloc[:, 0], errors='coerce')
    for col in table.columns[1:]:
        table[col] = pd.to_numeric(table[col], errors='coerce')

    plt.figure(figsize=(12, 8))
    for column in table.columns[1:]:
        plt.plot(table.iloc[:, 0], table[column], marker='o', label=column)

    plt.title(f"Creation Time vs Qubits - Table {i+1}")
    plt.xlabel("Number of Qubits")
    plt.ylabel("Creation Time (seconds)")
    plt.legend(title="Quantum Algorithms", bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```
